```python
from langchain_litellm import ChatLiteLLM
import os

LLM_MODEL_NAME = "litellm_proxy/openrouter/openai/gpt-4.1-mini"
# LLM_MODEL_NAME = "litellm_proxy/openrouter/openai/gpt-oss-120b"
# LLM_MODEL_NAME = "litellm_proxy/openrouter/anthropic/claude-haiku-4.5"
# LLM_MODEL_NAME = "litellm_proxy/openrouter/deepseek/deepseek-v3.2"
# LLM_MODEL_NAME = "litellm_proxy/openrouter/google/gemma-4-26b-a4b-it"

llm = ChatLiteLLM(
            model_name=LLM_MODEL_NAME,
            api_key=os.getenv("LITELLM_MASTER_KEY"),
            api_base=os.getenv("LITELLM_PROXY_URL"),
            request_timeout=120
        )
```

```python
response = llm.invoke("Hello, who are you?").content

response
```
